Descrição das tabelas:

1. DATA: Data da compra;
2. CO_ID: Identificação do número de compra (número da nota fiscal);

5. CL_EC: Estado civil do cliente:
1: Casado ou união estával;
2: Divorciado;

6. CL_FHL: Número de filhos do cliente;
7. CL_SEG: Segmentação econômica do cliente (classe A, B ou C);
8. PR_ID: Código do produto (SKU) adquirido;
9. PR_CAT: Categoria do produto adquirido;
10. PR_NOME: Nome do produto adquirido.

 Crie outro dataframe a partir de apenas as colunas mais importantes para poder trabalhar e analisar, escolhi: 'CO_ID','CL_ID','DATA','CL_GENERO', 'CL_EC', 'CL_FHL', 'PR_CAT', 'PR_NOME'

Para esta analise exploratoria de dados deste arquivo se tomaram apenas 100 linhas de registro. Para fazer para fines academicos e de aprendizado, pois o arquivo original contem mais de 1.000 linhas.

```python
df_enxuto= pd.read_csv('Base Varejo.csv', sep=";", encoding='latin-1',usecols=['CO_ID','CL_ID','DATA','CL_GENERO', 'CL_EC', 'CL_FHL', 'CL_SEG', 'PR_CAT', 'PR_NOME'], nrows=100) # crie o codigo completo padrão para leitura passando o arquivo original para um dataframe chamado df_enxuto com separado por ; encodinf latin-1
df_enxuto.head()
```

```python
df_enxuto.info()
```

Se realizou tratamento da coluna de CL_EC que referese a Estado civil do cliente para trocar os numeros para seus respetiva descrição

```python
df_enxuto['CL_EC']=df_enxuto['CL_EC'].replace(1, 'Casado ou união estável')
df_enxuto['CL_EC']=df_enxuto['CL_EC'].replace(2, 'Divorciado')
df_enxuto['CL_EC']=df_enxuto['CL_EC'].replace(3, 'Separado')
df_enxuto['CL_EC']=df_enxuto['CL_EC'].replace(4, 'Solteiro')
df_enxuto['CL_EC']=df_enxuto['CL_EC'].replace(5, 'Viúvo')
```
